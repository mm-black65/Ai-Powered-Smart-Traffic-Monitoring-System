```python
import cv2
```

```python
import numpy as np
```

```python
pts = np.array([
    [100,300],
    [250,100],
    [400,300]
], np.int32)
```

```python
canvas = np.ones((500,700,3), dtype=np.uint8) * 255
```

```python
poly = cv2.polylines(
    canvas,
    [pts],
    True,
    (255,255,255),
    3
)
cv2.imshow("Polygon", poly)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
poly = cv2.fillPoly(
    canvas,
    [pts],
    (255,0,255)
)
cv2.imshow("Polygon", poly)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
cc = cv2.putText(
    canvas,
    "Hello OpenCV",
    (50,450),
    cv2.FONT_HERSHEY_SIMPLEX,
    1,
    (255,0,255),
    2
)
cv2.imshow("text",cc)
cv2.waitKey(0)
cv2.destroyAllWindows()

#common font
""" 
cv2.FONT_HERSHEY_SIMPLEX

cv2.FONT_HERSHEY_COMPLEX

cv2.FONT_HERSHEY_DUPLEX

cv2.FONT_HERSHEY_SCRIPT_SIMPLEX"""
```

```python
def mouse(event, x, y, flags, param):
    if event == cv2.EVENT_LBUTTONDOWN:
        print(x, y)

cv2.namedWindow("Canvas")

cv2.setMouseCallback("Canvas", mouse)
```
